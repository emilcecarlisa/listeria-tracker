## About the USDA FSIS Sampling Data

### Data Source and Purpose

This analysis uses data from the **USDA Food Safety and Inspection Service (FSIS) Establishment-Specific Laboratory Sampling** program for Ready-to-Eat (RTE) products during Fiscal Year 2025 (October 1, 2024 - September 30, 2025).

According to USDA FSIS:

- **Purpose**: This sampling program monitors RTE meat and poultry products and food processing environments for pathogen contamination, particularly *Listeria monocytogenes* and *Salmonella*
- **Scope**: Samples are collected from federally-inspected meat and poultry establishments across all 50 states and U.S. territories
- **Sample Types**: Include both:
  - **Product samples**: Actual food products ready for consumption
  - **Environmental samples**: Food contact surfaces (equipment, conveyors, slicers) and non-food contact surfaces (floors, drains, walls)
- **Sampling Strategy**: Combines multiple approaches:
  - **Routine/Random sampling**: Regular surveillance at various establishments
  - **Risk-based sampling**: Targeted sampling based on risk algorithms
  - **Intensified sampling (IVT)**: For-cause testing at facilities with previous positive results

### Dataset Composition

- **Primary Dataset**: 27,211 sampling records with test results
- **Secondary Dataset**: 443 detailed genetic characterizations of positive samples
- **Coverage**: 2,364 unique food processing establishments
- **Geographic Scope**: All 50 states plus DC, Puerto Rico, and Guam

## Statistical Considerations & Limitations

### Understanding the Sampling Strategy

The USDA FSIS sampling program is **NOT purely random sampling**. It uses a hybrid approach:

1. **Random Component**: Some samples are collected as part of routine surveillance (RTEPROD program)
2. **Risk-Based Component**: Many samples are targeted based on:
   - Establishment history
   - Product type risk profiles
   - Previous positive results
   - Manufacturing process complexity
3. **For-Cause Component**: Intensified testing programs (IVT) specifically target facilities with previous contamination issues

### What This Means for Statistical Inference

#### ✅ What We CAN Conclude with Confidence:

Because the data includes both random and risk-based components:

- **Detection of contamination**: We can confidently identify which establishments, product types, and sample sources tested positive for pathogens
- **Comparative risk assessment**: We can compare contamination rates across different product categories, sample sources, and geographic regions *within this dataset*
- **Temporal patterns**: We can identify time-based trends in detection rates during FY2025
- **Effectiveness of interventions**: For facilities with repeated sampling, we can track whether positive rates decrease over time
- **Risk stratification**: We can identify product categories and sample source types with higher detection rates

#### ⚠️ What We SHOULD BE CAUTIOUS About:

- **Population-wide prevalence estimates**: Because sampling is not purely random, detection rates in this dataset may **overestimate** true contamination prevalence in the entire U.S. food supply
  - Reason: Risk-based and for-cause sampling intentionally targets higher-risk facilities
- **Generalization to untested facilities**: Results may not represent facilities that were not included in the sampling program
- **Selection bias**: Larger establishments and certain product categories may be sampled more frequently than others

#### ❌ What We CANNOT Conclude:

- **Causation**: We can observe associations (e.g., "Product type X has higher positive rates") but cannot determine causation without additional data
- **Consumer risk**: This data shows contamination at the manufacturing level, NOT:
  - Whether contaminated products reached consumers
  - Whether anyone became ill
  - The extent of product distribution
- **True population prevalence**: The 0.77% Listeria detection rate in this dataset does NOT mean 0.77% of all RTE products nationwide are contaminated
- **Absence of contamination**: A negative test result means the sample tested negative, not that the entire facility or production batch is pathogen-free

### Sample Size Considerations

With **27,211 total samples** and **208 Listeria positives**:

- We have adequate statistical power to detect patterns in frequently-sampled product categories
- For rare product types with only a handful of samples, detection rates have **wide confidence intervals** and should be interpreted cautiously
- Subgroup analyses (e.g., positives by product type) become less reliable as sample sizes decrease

### Confidence in Different Analysis Types

| Analysis Type | Confidence Level | Reasoning |
|--------------|------------------|----------|
| Overall detection rates in dataset | **High** | Large sample size, comprehensive coverage |
| Comparison of product categories | **Moderate to High** | Valid within dataset, but subject to sampling strategy bias |
| Geographic patterns | **Moderate** | State-level sampling may be influenced by number of establishments per state |
| Temporal trends (within FY2025) | **High** | Consistent sampling methodology throughout the year |
| Facility-specific risk assessment | **High** | Direct observation of facility-level results |
| Extrapolation to untested facilities | **Low** | Sampling is targeted, not representative |
| Consumer health risk | **Cannot assess** | No data on illness, product distribution, or exposure |

## If Samples Were Purely Random: What Could We Conclude?

If the USDA FSIS data represented **purely random sampling** across all establishments and products (which it does not), we could make stronger statistical inferences:

### Statistical Inference with Random Sampling

#### Population Prevalence Estimation

With random sampling, the observed 0.77% Listeria detection rate could be used to estimate true population prevalence with calculable confidence intervals:

- **Point estimate**: 0.77% contamination rate
- **95% Confidence Interval** (using normal approximation):
  - Formula: $\hat{p} \pm 1.96 \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$
  - With n=27,124 and p=0.0077: **0.66% to 0.88%**
  
This would mean we could say with 95% confidence that the true contamination rate in the population is between 0.66% and 0.88%.

#### Comparative Analysis Between Groups

We could use statistical tests (chi-square, z-tests) to determine if observed differences are statistically significant:

- **Between product categories**: "Are deli meats significantly more contaminated than pork sausages?"
- **Between geographic regions**: "Does California have a significantly different contamination rate than Texas?"
- **Between sampling periods**: "Is the contamination rate in winter significantly higher than summer?"

Results would generalize to the broader population of establishments and products.

#### Prediction and Risk Assessment

- Could build predictive models that apply to untested facilities
- Could estimate probability that a randomly selected product would test positive
- Could calculate expected number of contaminated products in market

### Why This Dataset Is Still Valuable (Despite Non-Random Sampling)

Even though the sampling is not purely random:

1. **Public health surveillance**: Identifies actual contamination events requiring intervention
2. **Risk-based resource allocation**: Shows where USDA should focus inspection efforts
3. **Facility-level tracking**: Monitors individual establishments over time
4. **Outbreak detection**: Genetic fingerprinting helps identify potential outbreak clusters
5. **Program effectiveness**: Evaluates whether targeted interventions reduce contamination

The data is fit-for-purpose for **regulatory surveillance and enforcement**, even if not ideal for population-level epidemiological inference.

## Analysis 1: Sample Source Distribution

First, we'll examine the distribution of sample sources to understand what types of samples were collected.

```python
# Count sample sources
sample_sources = Counter([r['sample_source_name'] for r in primary_data
                         if r['sample_source_name'] not in [None, 'NULL', '']])

print("="*80)
print("SAMPLE SOURCE DISTRIBUTION")
print("="*80)
print(f"\nTotal unique sample source types: {len(sample_sources)}")
print(f"\nTop 20 Most Common Sample Sources:")
print(f"{'Sample Source':<70} {'Count':>8} {'%':>6}")
print("-" * 85)

for source, count in sample_sources.most_common(20):
    pct = (count / len(primary_data)) * 100
    source_display = source[:67] + "..." if len(source) > 67 else source
    print(f"{source_display:<70} {count:>8,} {pct:>5.1f}%")
```

```python
# Categorize sample sources into Product vs Environmental
def categorize_sample_source(source_name):
    """
    Categorize sample source into Product, Food Contact Surface, or Non-Food Contact Surface
    """
    if source_name in [None, 'NULL', '']:
        return 'Unknown'
    
    source_lower = source_name.lower()
    
    # Non-product contact surfaces (floors, walls, drains)
    if any(keyword in source_lower for keyword in 
           ['non-product contact', 'non product contact', 'floor', 'wall', 'drain']):
        return 'Non-Food Contact Surface'
    
    # Food contact surfaces (equipment, sponges, swabs)
    if any(keyword in source_lower for keyword in 
           ['contact surface', 'sponge', 'swab', 'equipment']):
        return 'Food Contact Surface'
    
    # Everything else is considered a product sample
    return 'Product Sample'

# Apply categorization
sample_categories = defaultdict(lambda: {'total': 0, 'lm_positive': 0, 'sal_positive': 0})

for record in primary_data:
    source = record['sample_source_name']
    category = categorize_sample_source(source)
    
    sample_categories[category]['total'] += 1
    
    if record['lm_listeria_analysis'] == 'Positive':
        sample_categories[category]['lm_positive'] += 1
    
    if record['salmonella_sp_analysis'] == 'Positive':
        sample_categories[category]['sal_positive'] += 1

print("\n" + "="*80)
print("SAMPLE CATEGORY BREAKDOWN")
print("="*80)
print(f"\n{'Category':<30} {'Total':>10} {'Lm+ ':>8} {'Lm Rate':>10} {'Sal+':>8} {'Sal Rate':>10}")
print("-" * 80)

for category, stats in sorted(sample_categories.items(), key=lambda x: x[1]['total'], reverse=True):
    total = stats['total']
    lm_pos = stats['lm_positive']
    sal_pos = stats['sal_positive']
    lm_rate = (lm_pos / total * 100) if total > 0 else 0
    sal_rate = (sal_pos / total * 100) if total > 0 else 0
    
    print(f"{category:<30} {total:>10,} {lm_pos:>8} {lm_rate:>9.2f}% {sal_pos:>8} {sal_rate:>9.2f}%")
```

## Visualization 1: Sample Category Distribution

This visualization shows the breakdown of samples by category (Product vs Environmental) and their contamination rates.

```python
# Prepare data for visualization
categories = []
totals = []
lm_rates = []

for category, stats in sorted(sample_categories.items(), key=lambda x: x[1]['total'], reverse=True):
    if category != 'Unknown':
        categories.append(category)
        totals.append(stats['total'])
        lm_rate = (stats['lm_positive'] / stats['total'] * 100) if stats['total'] > 0 else 0
        lm_rates.append(lm_rate)

# Create figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Sample volume by category
colors = ['#3498db', '#e74c3c', '#2ecc71']
bars1 = ax1.bar(range(len(categories)), totals, color=colors[:len(categories)], alpha=0.8, edgecolor='black')
ax1.set_xlabel('Sample Category', fontsize=12, fontweight='bold')
ax1.set_ylabel('Number of Samples', fontsize=12, fontweight='bold')
ax1.set_title('Sample Volume by Category', fontsize=14, fontweight='bold')
ax1.set_xticks(range(len(categories)))
ax1.set_xticklabels(categories, rotation=15, ha='right')
ax1.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(height):,}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# Plot 2: Listeria positive rate by category
bars2 = ax2.bar(range(len(categories)), lm_rates, color=colors[:len(categories)], alpha=0.8, edgecolor='black')
ax2.set_xlabel('Sample Category', fontsize=12, fontweight='bold')
ax2.set_ylabel('Listeria Positive Rate (%)', fontsize=12, fontweight='bold')
ax2.set_title('Listeria Detection Rate by Category', fontsize=14, fontweight='bold')
ax2.set_xticks(range(len(categories)))
ax2.set_xticklabels(categories, rotation=15, ha='right')
ax2.grid(axis='y', alpha=0.3)
ax2.axhline(y=0.77, color='red', linestyle='--', linewidth=2, label='Overall Avg (0.77%)')
ax2.legend()

# Add value labels on bars
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.2f}%',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n📊 Key Observation: Environmental samples (contact surfaces) are tested more frequently than product samples.")
print("   This reflects USDA's strategy of monitoring facility sanitation to prevent product contamination.")
```

## Analysis 2: Product Type Contamination Rates

Now we'll analyze contamination rates for specific product types to identify high-risk categories.

```python
# Analyze product-specific contamination (only actual product samples, not environmental)
product_samples = [r for r in primary_data 
                  if categorize_sample_source(r['sample_source_name']) == 'Product Sample']

# Calculate statistics for each product type
product_stats = defaultdict(lambda: {'total': 0, 'lm_positive': 0, 'sal_positive': 0})

for record in product_samples:
    product_type = record['sample_source_name']
    if product_type not in [None, 'NULL', '']:
        product_stats[product_type]['total'] += 1
        
        if record['lm_listeria_analysis'] == 'Positive':
            product_stats[product_type]['lm_positive'] += 1
        
        if record['salmonella_sp_analysis'] == 'Positive':
            product_stats[product_type]['sal_positive'] += 1

# Filter products with at least 10 samples for statistical reliability
min_samples = 10
product_stats_filtered = {k: v for k, v in product_stats.items() if v['total'] >= min_samples}

# Calculate positive rates and sort
product_rates = []
for product, stats in product_stats_filtered.items():
    lm_rate = (stats['lm_positive'] / stats['total'] * 100) if stats['total'] > 0 else 0
    sal_rate = (stats['sal_positive'] / stats['total'] * 100) if stats['total'] > 0 else 0
    product_rates.append({
        'product': product,
        'total': stats['total'],
        'lm_positive': stats['lm_positive'],
        'lm_rate': lm_rate,
        'sal_positive': stats['sal_positive'],
        'sal_rate': sal_rate
    })

# Sort by Listeria positive rate
product_rates_sorted = sorted(product_rates, key=lambda x: x['lm_rate'], reverse=True)

print("="*100)
print("PRODUCT-SPECIFIC CONTAMINATION RATES")
print(f"(Products with at least {min_samples} samples)")
print("="*100)
print(f"\n{'Product Type':<50} {'Samples':>8} {'Lm+':>6} {'Lm Rate':>10} {'Sal+':>6} {'Sal Rate':>10}")
print("-" * 100)

# Show top 25 by Listeria rate
for item in product_rates_sorted[:25]:
    product_display = item['product'][:47] + "..." if len(item['product']) > 47 else item['product']
    print(f"{product_display:<50} {item['total']:>8} {item['lm_positive']:>6} "
          f"{item['lm_rate']:>9.2f}% {item['sal_positive']:>6} {item['sal_rate']:>9.2f}%")

print(f"\n\nTotal product types analyzed: {len(product_stats_filtered)}")
print(f"Total product samples: {len(product_samples):,}")
```

## Visualization 2: Top Product Categories by Contamination Rate

This shows the highest-risk product categories ranked by Listeria detection rate.

```python
# Visualize top products by contamination rate
top_n = 15
top_products = product_rates_sorted[:top_n]

# Extract data for plotting
product_names = [p['product'][:40] for p in top_products]  # Truncate long names
lm_rates_top = [p['lm_rate'] for p in top_products]
sample_counts = [p['total'] for p in top_products]

# Create figure
fig, ax = plt.subplots(figsize=(14, 10))

# Create horizontal bar chart
y_pos = np.arange(len(product_names))
colors_gradient = plt.cm.RdYlGn_r(np.linspace(0.2, 0.8, len(product_names)))

bars = ax.barh(y_pos, lm_rates_top, color=colors_gradient, alpha=0.8, edgecolor='black')

ax.set_yticks(y_pos)
ax.set_yticklabels(product_names, fontsize=10)
ax.set_xlabel('Listeria Detection Rate (%)', fontsize=12, fontweight='bold')
ax.set_title(f'Top {top_n} Product Types by Listeria Detection Rate\n(Minimum {min_samples} samples)', 
             fontsize=14, fontweight='bold')
ax.grid(axis='x', alpha=0.3)

# Add value labels
for i, (bar, rate, count) in enumerate(zip(bars, lm_rates_top, sample_counts)):
    width = bar.get_width()
    ax.text(width, bar.get_y() + bar.get_height()/2,
            f' {rate:.2f}% (n={count})',
            ha='left', va='center', fontsize=9, fontweight='bold')

# Add reference line for overall average
overall_product_lm_rate = (sum(p['lm_positive'] for p in product_rates) / 
                           sum(p['total'] for p in product_rates) * 100)
ax.axvline(x=overall_product_lm_rate, color='blue', linestyle='--', linewidth=2, 
          label=f'Product Avg ({overall_product_lm_rate:.2f}%)')
ax.legend(loc='lower right')

plt.tight_layout()
plt.gca().invert_yaxis()  # Highest rate at top
plt.show()

print(f"\n📊 Note: Sample size (n) shown for each product. Smaller sample sizes have wider confidence intervals.")
```

## Analysis 3: Environmental Contamination - Contact Surfaces

Environmental monitoring is critical for preventing product contamination. We'll analyze food contact and non-food contact surface results.

```python
# Analyze environmental samples specifically
environmental_samples = [r for r in primary_data 
                        if categorize_sample_source(r['sample_source_name']) in 
                        ['Food Contact Surface', 'Non-Food Contact Surface']]

# Break down by specific surface types
surface_stats = defaultdict(lambda: {'total': 0, 'lm_positive': 0})

for record in environmental_samples:
    surface_type = record['sample_source_name']
    if surface_type not in [None, 'NULL', '']:
        surface_stats[surface_type]['total'] += 1
        if record['lm_listeria_analysis'] == 'Positive':
            surface_stats[surface_type]['lm_positive'] += 1

# Filter surfaces with at least 20 samples
min_surface_samples = 20
surface_stats_filtered = {k: v for k, v in surface_stats.items() if v['total'] >= min_surface_samples}

# Calculate rates
surface_rates = []
for surface, stats in surface_stats_filtered.items():
    rate = (stats['lm_positive'] / stats['total'] * 100) if stats['total'] > 0 else 0
    surface_rates.append({
        'surface': surface,
        'total': stats['total'],
        'lm_positive': stats['lm_positive'],
        'rate': rate
    })

surface_rates_sorted = sorted(surface_rates, key=lambda x: x['rate'], reverse=True)

print("="*90)
print("ENVIRONMENTAL SURFACE CONTAMINATION RATES")
print(f"(Surfaces with at least {min_surface_samples} samples)")
print("="*90)
print(f"\n{'Surface Type':<60} {'Samples':>10} {'Lm+':>8} {'Rate':>10}")
print("-" * 90)

for item in surface_rates_sorted[:20]:
    surface_display = item['surface'][:57] + "..." if len(item['surface']) > 57 else item['surface']
    print(f"{surface_display:<60} {item['total']:>10,} {item['lm_positive']:>8} {item['rate']:>9.2f}%")
```

## Conclusions & Caveats

### ✅ What We Can Conclude with Confidence:

1. **Environmental Monitoring Strategy**: The USDA FSIS program heavily emphasizes environmental sampling (food contact and non-food contact surfaces), which accounts for the majority of samples. This is a proactive contamination prevention strategy.

2. **Relative Risk Assessment**: Within this dataset, we can confidently rank product types and sample sources by their observed contamination rates. This helps prioritize inspection resources.

3. **Environmental vs Product Contamination**: Environmental samples show higher detection rates than finished product samples, suggesting that facility sanitation monitoring successfully identifies contamination before it reaches the final product.

4. **Product-Specific Patterns**: Certain product categories (visible in the analysis) show elevated detection rates and warrant closer monitoring.

### ⚠️ Important Caveats:

1. **Sampling Bias**: The data includes risk-based and for-cause sampling, which intentionally targets higher-risk facilities. Therefore:
   - Detection rates shown here likely **overestimate** the true contamination prevalence in the broader food supply
   - Products with higher detection rates may be sampled more frequently due to perceived risk
   - Facilities with previous positives receive intensified testing

2. **Sample Size Variation**: Products with small sample sizes (n < 50) have wide confidence intervals. A single positive sample out of 10 yields a 10% detection rate, but this is not statistically robust.

3. **Geographic and Temporal Limitations**: This analysis does not account for:
   - Geographic clustering of facilities
   - Seasonal contamination patterns
   - Changes in sampling protocols over time

4. **Consumer Risk**: This data shows contamination at the manufacturing level, **NOT**:
   - Whether contaminated products reached consumers
   - Whether anyone became ill
   - The effectiveness of recalls or corrective actions
